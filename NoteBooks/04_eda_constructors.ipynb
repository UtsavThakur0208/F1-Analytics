```python
# ============================================================
# NOTEBOOK 4: Constructor EDA
# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

master = pd.read_csv("/content/drive/MyDrive/f1-analytics/data/processed/processedmaster_df.csv")
constructor_stats = pd.read_csv("/content/drive/MyDrive/f1-analytics/data/processed/processedconstructor_season_stats.csv")

master_modern = master[master['year'] >= 2000].copy()
constructor_stats_modern = constructor_stats[constructor_stats['year'] >= 2000].copy()
```

```python
# ── ANALYSIS 1: CONSTRUCTOR WIN DOMINANCE ─────────────────────

constructor_wins = (
    master_modern[master_modern['is_winner'] == 1]
    .groupby('constructor_name')['is_winner']
    .sum()
    .sort_values(ascending=True)
    .tail(12)
    .reset_index()
)
constructor_wins.columns = ['Constructor', 'Wins']

fig5 = px.bar(
    constructor_wins,
    x='Wins',
    y='Constructor',
    orientation='h',
    title='Constructor Wins — 2000 to Present',
    color='Wins',
    color_continuous_scale='Blues',
    text='Wins'
)
fig5.update_layout(plot_bgcolor='white', paper_bgcolor='white', coloraxis_showscale=False)
fig5.update_traces(textposition='outside')
fig5.show()
```

```python
# ── ANALYSIS 2: CONSTRUCTOR POINTS TREND ──────────────────────
#
# WHY: Shows dynasty cycles. Ferrari dominated 2000-2004,
# then Red Bull 2010-2013, Mercedes 2014-2021, Red Bull again.
# This narrative makes your project feel like real analysis.

top_constructors = (
    constructor_stats_modern.groupby('constructor_name')['total_points']
    .sum()
    .nlargest(8)
    .index.tolist()
)

trend_data = constructor_stats_modern[
    constructor_stats_modern['constructor_name'].isin(top_constructors)
]

fig6 = px.line(
    trend_data,
    x='year',
    y='total_points',
    color='constructor_name',
    title='Constructor Points Trend — Top 8 Teams (2000-Present)',
    markers=True,
    labels={'total_points': 'Season Points', 'constructor_name': 'Team'}
)
fig6.update_layout(plot_bgcolor='white', paper_bgcolor='white')
fig6.show()
```

```python
# ── ANALYSIS 3: RELIABILITY — DNF RATE BY CONSTRUCTOR ─────────
#
# WHY: A team that's fast but unreliable won't win championships.
# DNF rate is a real engineering KPI used in motorsport.
# This shows you understand the BUSINESS of F1, not just results.

reliability = constructor_stats_modern.groupby('constructor_name').agg(
    total_starts = ('total_races', 'sum'),
    total_dnf    = ('total_dnf', 'sum')
).reset_index()

reliability = reliability[reliability['total_starts'] >= 100]  # enough data
reliability['dnf_rate'] = (reliability['total_dnf'] / reliability['total_starts'] * 100).round(1)
reliability = reliability.sort_values('dnf_rate')

fig7 = px.bar(
    reliability,
    x='dnf_rate',
    y='constructor_name',
    orientation='h',
    title='Constructor DNF Rate (%) — Lower is Better (Min 100 Starts)',
    color='dnf_rate',
    color_continuous_scale='RdYlGn_r',  # red=bad, green=good
    text='dnf_rate'
)
fig7.update_layout(plot_bgcolor='white', paper_bgcolor='white')
fig7.show()
```
